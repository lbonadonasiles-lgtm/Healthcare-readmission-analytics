```python
import pandas as pd
import sqlite3
import os

project_path = os.path.join(os.path.expanduser("~"), "Documents", "Healthcare_SQL_Analytics")

data_path = os.path.join(
    project_path,
    "data",
    "hospital_readmissions.csv"
)

db_path = os.path.join(
    project_path,
    "healthcare_readmissions.db"
)

print("Project path:", project_path)
print("Data path:", data_path)
print("Database path:", db_path)

print("\nData file exists:", os.path.exists(data_path))
```

```python
df = pd.read_csv(data_path)

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

df.head()
print("Column names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())
```

```python
for col in df.select_dtypes(include="object").columns:
    print(f"\n--- {col} ---")
    print(df[col].value_counts())
```

```python
print(df["readmitted"].value_counts())

print("\nReadmission Rate:")
print(round(df["readmitted"].value_counts(normalize=True) * 100, 2))
```

```python
pd.crosstab(
    df["change"],
    df["readmitted"],
    normalize="index"
) * 100
```

```python
df.groupby("readmitted")["n_inpatient"].mean()
```

```python
df.groupby("readmitted")["n_emergency"].mean()
```

```python
df.groupby("readmitted")["time_in_hospital"].mean()
```
